```python
import numpy as np
import matplotlib.pyplot as plt

# Definición de la EDO: dy/dt = 2 * t * y
def f(t, y):
    return 2 * t * y

# Solución analítica exacta: y(t) = e^(t^2)
def solucion_exacta(t):
    return np.exp(t**2)

# Parámetros del problema
t0 = 0.0
tf = 1.0
y0 = 1.0
h = 0.2

# Método de Euler
n_pasos = int((tf - t0) / h)
t_valores = [t0]
y_euler = [y0]

t_actual = t0
y_actual = y0

for _ in range(n_pasos):
    y_actual = y_actual + h * f(t_actual, y_actual)
    t_actual = round(t_actual + h, 10)
    t_valores.append(t_actual)
    y_euler.append(y_actual)

t_valores = np.array(t_valores)
y_euler = np.array(y_euler)
y_exacta = solucion_exacta(t_valores)
error_abs = np.abs(y_exacta - y_euler)

# Tabla comparativa en consola
print(f"{'t':>6} | {'y_exacta':>12} | {'y_euler':>12} | {'Error Abs':>12}")
print("-" * 50)
for t, ye, yeu, err in zip(t_valores, y_exacta, y_euler, error_abs):
    print(f"{t:6.1f} | {ye:12.6f} | {yeu:12.6f} | {err:12.6f}")

# Gráfica comparativa
t_fino = np.linspace(t0, tf, 100)
plt.figure(figsize=(8, 5))
plt.plot(t_fino, solucion_exacta(t_fino), label="Solución Analítica: $y(t) = e^{t^2}$", linewidth=2)
plt.plot(t_valores, y_euler, "o--", label="Aproximación de Euler (h = 0.2)", markersize=6)
plt.title("Comparación: Solución Analítica vs. Método de Euler")
plt.xlabel("t")
plt.ylabel("y(t)")
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend()
plt.tight_layout()
plt.savefig("comparacion_euler.png", dpi=150)
plt.show()
```
